```python
import os
import shutil
import kagglehub
import pandas as pd

# 1. Скачиваем датасет через базовый метод kagglehub
dataset_cache_dir = kagglehub.dataset_download("mlg-ulb/creditcardfraud")

# 2. Определяем путь к локальной папке
local_dir = "../data/raw/"
os.makedirs(local_dir, exist_ok=True)

# 3. Переносим файл в рабочую директорию
local_file_path = os.path.join(local_dir, "creditcard.csv")
shutil.copy(os.path.join(dataset_cache_dir, "creditcard.csv"), local_file_path)

# 4. Читаем уже напрямую из локальной папки
df = pd.read_csv(local_file_path)
print("First 5 records:\n", df.head())
```
