```python
import os
import shutil
import kagglehub
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 1. Скачиваем датасет через базовый метод kagglehub
dataset_cache_dir = kagglehub.dataset_download("mlg-ulb/creditcardfraud")

# 2. Определяем путь к локальной папке
local_dir = "../data/raw/"
os.makedirs(local_dir, exist_ok=True)

# 3. Переносим файл в рабочую директорию
local_file_path = os.path.join(local_dir, "creditcard.csv")
shutil.copy(os.path.join(dataset_cache_dir, "creditcard.csv"), local_file_path)

# 4. Читаем уже напрямую из локальной папки
fraud_df = pd.read_csv(local_file_path)
print("First 5 records:\n", fraud_df.head())
```

```python
# Set visual style for publication
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di 
# 1. Data Preparation
# Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di 
fraud_counts = fraud_df["Class"].value_counts()
fraud_percent = (fraud_counts[1] / len(fraud_df)) * 100
total_loss = fraud_df[fraud_df["Class"] == 1]["Amount"].sum()

# Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di 
# Chart 1: Transaction Class Imbalance
# Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di 
bars = axes[0].bar(["Legitimate (0)", "Fraudulent (1)"], fraud_counts.values, color=["#2b5c8f", "#e74c3c"])
axes[0].set_yscale("log")
axes[0].set_title(f"Class Imbalance\n(Fraud: {fraud_percent:.2f}% | Total Loss: ${total_loss:,.2f})", fontsize=12, fontweight="bold")
axes[0].set_ylabel("Count (log scale)")

for bar in bars:
    height = bar.get_height()
    axes[0].annotate(f"{height:,}",
                xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 3),  
                textcoords="offset points",
                ha="center", va="bottom")

# Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di 
# Chart 2: Average Transaction Amount
# Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di 
sns.barplot(x="Class", y="Amount", data=fraud_df, ax=axes[1], palette=["#2b5c8f", "#e74c3c"], ci=None)
axes[1].set_xticks([0, 1])
axes[1].set_xticklabels(["Legitimate", "Fraudulent"])
axes[1].set_title("Average Transaction Amount ($)", fontsize=12, fontweight="bold")
axes[1].set_xlabel("Transaction Type")
axes[1].set_ylabel("Average Amount ($)")

mean_norm = fraud_df[fraud_df["Class"] == 0]["Amount"].mean()
mean_fraud = fraud_df[fraud_df["Class"] == 1]["Amount"].mean()

# Extend Y-axis upper limit to prevent text clipping
axes[1].set_ylim(0, max(mean_norm, mean_fraud) * 1.15)

axes[1].text(0, mean_norm + 2, f"${mean_norm:.2f}", ha="center", fontweight="bold")
axes[1].text(1, mean_fraud + 2, f"${mean_fraud:.2f}", ha="center", fontweight="bold")

# Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di 
# Chart 3: Temporal Fraud Distribution
# Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di 
fraud_df["Hour"] = (fraud_df["Time"] / 3600) % 48

sns.histplot(data=fraud_df[fraud_df["Class"] == 1], x="Hour", bins=48, ax=axes[2], color="#e74c3c", kde=True)
axes[2].set_title("Fraud Peaks by Hour (48-Hour Window)", fontsize=12, fontweight="bold")
axes[2].set_xlabel("Hour Since Start (0 - 48h)")
axes[2].set_ylabel("Fraudulent Transaction Count")
axes[2].axvline(x=24, color="black", linestyle="--", label="End of Day 1")
axes[2].legend()

# Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di 
# Render and Save
# Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di Di 
plt.tight_layout()
plt.savefig("../docs/fraud_analysis_charts.png")
plt.show()
```

## Exploratory Data Analysis: Credit Card Fraud Findings

### 1. Class Imbalance & Financial Loss
* **Type:** Logarithmic Bar Chart
* **Key Finding:** Demonstrates an extreme class imbalance within the dataset. Fraudulent transactions account for only **0.17%** of total activity (492 cases vs. 284,315 legitimate transactions).
* **Business Insight:** Despite the minor percentage, these transactions incurred a direct total loss of **$60,127.97**. A logarithmic scale is applied to make the fraudulent class visually interpretable alongside majority instances.

---

### 2. Transaction Amount Comparison (Mean Check)
* **Type:** Comparative Bar Chart
* **Key Finding:** The average value of a fraudulent transaction is **$122.21**, compared to **$88.29** for legitimate transactions.
* **Business Insight:** Fraudulent operations consistently target higher transaction amounts. This metric serves as a foundational baseline for establishing automated freeze thresholds and rule-based risk filters in anti-fraud monitoring systems.

---

### 3. Temporal Fraud Distribution (48-Hour Window)
* **Type:** Histogram with Kernel Density Estimate (KDE)
* **Key Finding:** Displays distinct time-series peaks in fraudulent activity across a 48-hour observation timeframe.
* **Business Insight:** Fraudulent operations exhibit pronounced temporal clustering during off-peak hours (late night/early morning hours of Day 1 and Day 2). This behavioral pattern reflects automated batch testing by fraudsters when cardholders are less likely to notice immediate push notifications.
